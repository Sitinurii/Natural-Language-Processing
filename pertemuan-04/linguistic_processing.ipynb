```python
# 2. Memuat Model spaCy
import spacy

nlp = spacy.load("model_indonesia")

print("Model spaCy Bahasa Indonesia berhasil dimuat!")
```

```python
# 3. Memproses Kalimat

doc = nlp("Mahasiswa belajar Natural Language Processing.")

for token in doc:
    print(f"{token.text:<15} {token.pos_:<10} {token.dep_}")
```

```python
# 4. Analisis POS Token

doc = nlp("Mahasiswa belajar Natural Language Processing.")

print(f"{'Token':<15} {'POS':<10} {'Lemma'}")
print("-" * 40)

for token in doc:
    print(f"{token.text:<15} {token.pos_:<10} {token.lemma_}")

# Catatan: kolom Lemma kosong karena modelm belum memiliki komponen lemmatizer.
#  Itu keterbatasan model, bukan kesalahan kode.
```

```python
# 5. Membuat Tabel Analisis POS

import pandas as pd

def analyze_pos(text, nlp_model):
    doc = nlp_model(text)
    results = []

    for token in doc:
        results.append({
            "Token": token.text,
            "POS": token.pos_,
            "POS Tag": token.tag_,
            "Lemma": token.lemma_,
            "Dependency": token.dep_,
            "Head": token.head.text
        })

    return pd.DataFrame(results)

df_pos = analyze_pos("Mahasiswa belajar NLP.", nlp)

print(df_pos.to_string(index=False))

# Catatan: POS Tag bernilai `X` dan Lemma kosong karena keterbatasan model yang digunakan.
```

```python
# 6. Analisis Beberapa Kalimat
sentences = [
    "Mahasiswa belajar NLP.",
    "Dosen mengajar Natural Language Processing.",
    "Mahasiswa membaca buku.",
    "Sistem memproses data teks.",
    "Model bahasa memprediksi kata berikutnya."
]

for sentence in sentences:
    print(f"\nKalimat: {sentence}")
    df_result = analyze_pos(sentence, nlp)
    print(df_result[["Token", "POS", "Dependency", "Head"]].to_string(index=False))
    print()
```

```python
# 7. Dependency Parsing 
doc = nlp("Mahasiswa belajar NLP.")

print(f"{'Token':<15} {'Dep':<12} {'Head'}")
print("-" * 40)
for token in doc:
    print(f"{token.text:<15} {token.dep_:<12} {token.head.text}")
```

```python
# 8. Visualisasi Dependency 
from spacy import displacy

doc = nlp("Mahasiswa belajar Natural Language Processing.")

displacy.render(
    doc,
    style="dep",
    jupyter=True,
    options={"distance": 90}
)
```

```python
# 9. Analisis Dataset

import pandas as pd

df = pd.read_csv("../pertemuan-03/dataset_preprocessed.csv")
text_column = "content"

# Mengambil data yang teksnya tidak kosong
df = df.dropna(subset=[text_column])
df = df[df[text_column].astype(str).str.strip() != ""]

# Mengambil 10 sampel
sample = df.sample(n=10, random_state=42)

for _, row in sample.iterrows():
    text = str(row[text_column])[:200]
    doc = nlp(text)

    pos_counts = {}

    for token in doc:
        pos = token.pos_
        pos_counts[pos] = pos_counts.get(pos, 0) + 1

    print(f"Teks: {text[:80]}...")
    print(f"POS Distribution: {pos_counts}")
    print()

    # Catatan: Analisis dilakukan pada 10 sampel ulasan Gojek.
    # Hasil POS menunjukkan bahwa setiap ulasan memiliki jenis kata yang berbeda. 
    # Namun, beberapa label mungkin kurang tepat karena keterbatasan model dalam memahami bahasa informal.
```

```python
# 10. Analisis Kesalahan
# Linguistic processing tidak selalu sempurna. Menguji dengan teks informal:
informal_texts = [
    "aku lg belajar NLP nih",
    "gue udah baca buku itu",
    "modelnya bagus bgt sih",
    "nggak ngerti sama sekali",
    "udah instal belum?"
]

print("Analisis Teks Informal:")
print("=" * 60)
for text in informal_texts:
    doc = nlp(text)
    print(f"\nTeks: {text}")
    for token in doc:
        print(f"  {token.text:<15} {token.pos_}")

# Catatan: Model masih mengalami kesalahan dalam mengenali beberapa kata informal dan singkatan, seperti “gue”, “nggak”, dan “bgt”. 
# Hal ini menunjukkan bahwa model memiliki keterbatasan dalam menganalisis bahasa sehari-hari.
```

```python
# E. Tugas Analisis Linguistik

import pandas as pd
from spacy import displacy

# 1. Memilih 10 kalimat dari dataset
kalimat_list = [
    "APLIKASI RIBETE SETENGAH MODARRRR JEKKK GOJEKKKKKK APLIKASI PEKOK KOK DI PELIHARA BUBARIN SAJA MAU NGOJEK AJA RIBETE KAYAK MAU URUS CERAI APLIKASI BODOHH",
    "Ok ok",
    "Suka banget promonya banyak",
    "Yg platinum selalu di kejar2 orderan Gmna mau trn levelnya Yg besik gmna bisa naik Orderannya Yg msk gado2 1 hari klo di ambil smua cuma msk 5 orderan Ky di suruh solat 5 waktu Yg platinum boro2 inget itungopi aja ga sempet Tlg di pertimbangkan lg Bonus hapus aja Kita dah rela Yg penting orderan merata Mana salam 1 aspal Kesejahteraan mana Dlu bp nadim waktu trening ngomongnya bgtu Trima ksh",
    "Sangat membantu",
    "Plissss jangan ad update lagi dong gojek",
    "Bagus",
    "Selalu ontimeramahgak pernah kesasar",
    "Persyaratan go pay plus RUWET Nyesel pake apk ini",
    "Bagus banget"
]

# 2. Tokenization, POS Tagging, dan Dependency Parsing
hasil = []

for nomor, kalimat in enumerate(kalimat_list, start=1):
    doc = nlp(kalimat)

    for token in doc:
        hasil.append({
            "Kalimat": kalimat,
            "Token": token.text,
            "POS": token.pos_,
            "Dependency": token.dep_,
            "Head": token.head.text
        })

# 3. Membuat tabel analisis
df_tugas = pd.DataFrame(hasil)

# Menggabungkan hasil token, POS, dan dependency per kalimat
hasil_ringkas = []

for kalimat, kelompok in df_tugas.groupby("Kalimat", sort=False):
    hasil_ringkas.append({
        "Kalimat": kalimat,
        "Token": ", ".join(kelompok["Token"].astype(str)),
        "POS": ", ".join(
            f"{token}: {pos}"
            for token, pos in zip(
                kelompok["Token"], kelompok["POS"]
            )
        ),
        "Dependency": "; ".join(
            f"{token} → {head} ({dep})"
            for token, head, dep in zip(
                kelompok["Token"],
                kelompok["Head"],
                kelompok["Dependency"]
            )
        ),
        "Kesalahan?": "Periksa kesesuaian hasil model"
    })

df_ringkas = pd.DataFrame(hasil_ringkas)

print("TABEL HASIL ANALISIS LINGUISTIK")
display(df_ringkas)

# 4. Visualisasi Dependency Tree setiap kalimat
for nomor, kalimat in enumerate(kalimat_list, start=1):
    doc = nlp(kalimat)

    print(f"Dependency Tree Kalimat {nomor}")
    displacy.render(
        doc,
        style="dep",
        jupyter=True,
        options={"compact": True}
    )
```
