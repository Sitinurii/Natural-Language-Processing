```python
# 2. Import Library
import pandas as pd
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
# 2. Load Dataset
df = pd.read_csv("../dataset.csv")
text_column = "content"

df.head(5)
```

```python
# 4. Sentence Segmentation
text = "Saya belajar NLP. NLP sangat menarik. Saya ingin menjadi NLP Engineer."

sentences = sent_tokenize(text)

for sentence in sentences:
    print(sentence)
```

```python
# 5. Word Tokenization
from nltk.tokenize import word_tokenize

text = "Saya belajar Natural Language Processing."

tokens = word_tokenize(text)

print(tokens)
```

```python
# 6. Character Tokenization
text = "NLP"

characters = list(text)

print(characters)
```

```python
# 7. Tokenisasi Dataset
df = df.dropna(subset=[text_column]).copy()

df["tokens"] = df[text_column].astype(str).apply(word_tokenize)

df[[text_column, "tokens"]].head()
```

```python
df["jumlah_token"] = df["tokens"].apply(len)
df["jumlah_token"].describe()
```

```python
# 8. Membandingkan Kata dan Token
# Menghitung jumlah kata dengan split sederhana
df["jumlah_kata"] = df['content'].astype(str).apply(
    lambda x: len(x.split())
)

df[["jumlah_kata", "jumlah_token"]].describe()
```

```python
# 9. Eksperimen Tokenisasi
texts = [
    "Saya suka produk ini.",
    "Bagus banget!!!",
    "Tidak bagus...",
    "Harga Rp50.000.",
    "email@example.com",
    "https://example.com",
    "yg bagus bgt sih",
    "aku udah beli nih kak"
]

for text in texts:
    print("TEXT :", text)
    print("TOKEN:", word_tokenize(text))
    print()
```

```python
# 10. Perbandingan Tokenizer
import spacy

# Buat tokenizer multilingual tanpa model tambahan
nlp = spacy.blank("xx")

teks = "Saya belajar NLP."

# Tokenizer sederhana: memisahkan berdasarkan spasi
token_sederhana = teks.split()

# Tokenizer spaCy
doc = nlp(teks)
token_spacy = [token.text for token in doc]

print("Tokenizer sederhana:", token_sederhana)
print("Tokenizer spaCy    :", token_spacy)
```

```python
import nltk
import pandas as pd
from nltk.tokenize import word_tokenize

# Dibutuhkan oleh beberapa versi NLTK
nltk.download("punkt_tab", quiet=True)

teks_list = [
    "Saya belajar NLP.",
    "Bagus banget!!!",
    "Harga Rp50.000",
]

tabel = pd.DataFrame({
    "Teks": teks_list,
    "NLTK word_tokenize": [word_tokenize(teks) for teks in teks_list],
    "split()": [teks.split() for teks in teks_list],
})

display(tabel)
```

```python
# E. Tugas Analisis
1. Apakah setiap kata selalu menghasilkan satu token?
Tidak selalu. Satu kata bisa menjadi beberapa token, tergantung cara kerja tokenizer. Tanda baca juga dapat dihitung sebagai token tersendiri.
2. Bagaimana tokenizer menangani tanda baca?
Tokenizer dapat memisahkan tanda baca seperti titik, koma, dan tanda seru menjadi token tersendiri. Contohnya, Bagus banget!!! dapat dipecah menjadi ['Bagus', 'banget', '!', '!', '!'].
3. Bagaimana tokenizer menangani angka (contoh: Rp50.000)?
Hasilnya bergantung pada tokenizer yang digunakan. Pada NLTK word_tokenize, angka dan tanda titik dapat dipisahkan menjadi beberapa token.
4. Bagaimana tokenizer menangani URL?
URL dapat dipertahankan sebagai satu token atau dipecah menjadi beberapa bagian, tergantung tokenizer dan bentuk URL-nya.
5. Bagaimana tokenizer menangani singkatan informal (yg, dgn, kak)?
Tokenizer biasanya menganggap singkatan tersebut sebagai token biasa. Tokenizer tidak otomatis mengubah yg menjadi yang atau dgn menjadi dengan.
6. Apa masalah tokenisasi yang spesifik untuk Bahasa Indonesia?
Bahasa Indonesia memiliki banyak variasi bahasa informal, imbuhan, singkatan, campuran bahasa daerah dan asing, serta penggunaan tanda baca berulang. Hal ini dapat membuat proses tokenisasi lebih menantang.
7. Mengapa subword tokenizer diperlukan pada model Transformer?
Subword tokenizer memecah kata menjadi bagian yang lebih kecil sehingga model dapat menangani kata yang jarang muncul atau kata baru. Cara ini juga membantu model menggunakan kosakata secara lebih efisien.

Catatan: Jawaban ini berdasarkan konsep NLP dan eksperimen yang ada di modul. Untuk nomor 3 dan 4, hasil tokenisasi yang tepat sebaiknya dilihat langsung dari output eksperimen NLTK kamu.
```

```python
# G. Pertanyaan Diskusi
1. Apakah tokenizer yang baik untuk Bahasa Inggris otomatis baik untuk Bahasa Indonesia?
Tidak selalu, karena setiap bahasa memiliki aturan dan ciri yang berbeda. Bahasa Indonesia memiliki banyak imbuhan dan singkatan sehari-hari, sehingga tokenizer perlu disesuaikan dengan bahasa yang digunakan.
2. Apa yang terjadi jika teks tidak di-tokenize terlebih dahulu?
Teks akan lebih sulit diproses oleh sistem NLP karena belum dipecah menjadi bagian-bagian kecil, seperti kata atau subword. Tokenisasi membantu sistem memahami dan mengolah teks dengan lebih teratur.
3. Mengapa tanda baca bisa menjadi token tersendiri?
Karena tanda baca memiliki fungsi dalam kalimat, misalnya titik untuk mengakhiri kalimat, tanda tanya untuk pertanyaan, dan tanda seru untuk menunjukkan penekanan.
4. Apa hubungan tokenization dengan vocabulary dalam model NLP?
Tokenisasi berfungsi memecah teks menjadi token, sedangkan vocabulary merupakan kumpulan token yang dikenal oleh model. Token tersebut kemudian diubah menjadi angka agar dapat diproses oleh komputer.
5. Mengapa model Transformer menggunakan subword tokenization?
Karena subword tokenization dapat memecah kata menjadi bagian yang lebih kecil. Dengan cara ini, model bisa menangani kata yang jarang ditemukan atau kata baru tanpa harus menyimpan semua kata sebagai token yang berbeda.
```
