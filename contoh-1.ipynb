```python
# Menampilkan informasi dataset
print(data.info())

# Menampilkan ukuran dataset
print(data.shape)
```

```python
# Menampilkan jumlah dokumen
jumlah_dokumen = len(data)
print("Jumlah dokumen:", jumlah_dokumen)
```

```python
# Mengambil kolom teks ulasan
text_kolom = "content"
texts = data[text_kolom]
print("Contoh teks ulasan:", texts.head(10))
```

```python
# Melihat contoh teks ulasan
if "data" not in globals():
    import pandas as pd
    data = pd.read_csv("dataset.csv")

if "text_kolom" not in globals():
    text_kolom = "content"

if "texts" not in globals():
    texts = data[text_kolom].fillna("").astype(str)

for text in texts.head(5):
    print(text)
    print("-" * 50)
```

```python
# Memeriksa missing value
data.isnull().sum()
```

```python
# Menghitung missing value pada kolom content
missing_content = data[text_kolom].isnull().sum()
print("Jumlah missing value pada kolom content:", missing_content)
```

```python
# Menghapus data kosong
data = data.dropna(subset=[text_kolom])
print("Jumlah dokumen sebelum menghapus data kosong:", jumlah_dokumen)
print("Jumlah dokumen setelah menghapus data kosong:", len(data))
```

```python
# Memeriksa duplikasi
print("Jumlah dokumen duplikat:", data.duplicated().sum())
```

```python
# Menhapus duplikasi
data = data.drop_duplicates()
print("Jumlah dokumen sebelum menghapus duplikasi:", jumlah_dokumen)
print("Jumlah dokumen setelah menghapus duplikasi:", len(data))
```

```python
# Menghitung jumlah karakter pada isi ulasan
data["jumlah_karakter"] = data[text_kolom].fillna("").astype(str).apply(len)
data[[text_kolom, "jumlah_karakter"]].head(5)
```

```python
# Menghitung jumlah kata pada isi ulasan
data["jumlah_kata"] = data[text_kolom].fillna("").astype(str).apply(lambda x: len(x.split()))
data[[text_kolom, "jumlah_kata"]].head(7)
```

```python
# Menghitung jumlah kata pada isi ulasan
data["jumlah_kata"] = data[text_kolom].fillna("").astype(str).apply(lambda x: len(x.split()))
data[[text_kolom, "jumlah_kata"]].head(7)
```

```python
# Statistik panjang teks ulasan
print("Rata-rata jumlah kata:", data["jumlah_kata"].mean())
print("Jumlah kata minimum:", data["jumlah_kata"].min())
print("Jumlah kata maksimum:", data["jumlah_kata"].max())
print("Jumlah kata median:", data["jumlah_kata"].median())
print("Jumlah kata quartile 25%:", data["jumlah_kata"].quantile(0.25))
print("Jumlah kata quartile 75%:", data["jumlah_kata"].quantile(0.75))
```

```python
# 16. Menghitung jumlah kalimat dalam dataset
# Menghitung pada kolom content karena itu adalah teks ulasan

from nltk.tokenize import sent_tokenize

data["jumlah_kalimat"] = data[text_kolom].fillna("").astype(str).apply(
    lambda teks: len(sent_tokenize(teks))
)

print(data[[text_kolom, "jumlah_kalimat"]].head(10))
```

```python
# 17. Statistik corpus
print("Jumlah dokumen:", len(data))
print("Jumlah kata:", data["jumlah_kata"].sum())
print("Jumlah kalimat:", data["jumlah_kalimat"].sum())
print("Jumlah karakter:", data["jumlah_karakter"].sum())
```

```python
# Visualisasi distribusi panjang dokumen
import matplotlib.pyplot as plt

print("Grafik X: Distribusi Panjang Dokumen")
print("Sumbu X menunjukkan jumlah kata dalam setiap ulasan.")
print("Sumbu Y menunjukkan jumlah ulasan pada rentang jumlah kata tersebut.")

plt.figure(figsize=(10, 6))
plt.hist(data["jumlah_kata"], bins=30)
plt.xlabel("Jumlah Kata per Ulasan")
plt.ylabel("Jumlah Dokumen/Ulasan")
plt.title("Distribusi Panjang Dokumen")
plt.show()
```

```python
# 19. Visualisasi distribusi label/rating
import matplotlib.pyplot as plt

print("Grafik Y: Distribusi Label/Rating")
print("Sumbu X menunjukkan nilai score/rating dari setiap ulasan.")
print("Sumbu Y menunjukkan jumlah ulasan pada setiap nilai score.")

score_counts = data["score"].value_counts().sort_index()
print(score_counts)
score_counts.plot(kind="bar")
plt.xlabel("Score/Rating")
plt.ylabel("Jumlah Data/Ulasan")
plt.title("Distribusi Label/Rating")
plt.show()
```
